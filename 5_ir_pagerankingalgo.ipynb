```python
# Experiment: Implement PageRank Algorithm using NetworkX

import networkx as nx
import matplotlib.pyplot as plt

# Step 1: Create a directed graph
G = nx.DiGraph()

G.add_edges_from([
    (0, 1), (0, 2),
    (1, 2), (1, 3),
    (2, 0), (2, 3),
    (3, 0), (3, 1),
    (4, 2), (4, 3)
])

# Step 2: Calculate PageRank scores
pr = nx.pagerank(G, alpha=0.85)

# Step 3: Display PageRank scores
print("PageRank Scores:")

for node, score in pr.items():
    print(f"Node {node}: {score:.6f}")

# Step 4: Find the highest-ranked node
highest_rank = max(pr, key=pr.get)

print("\nNode with highest PageRank:", highest_rank)
print("Highest PageRank score:", round(pr[highest_rank], 6))

# Step 5: Rank nodes in descending order
print("\nNodes ranked from highest to lowest:")

ranked_nodes = sorted(
    pr.items(), key=lambda item: item[1], reverse=True
)

for node, score in ranked_nodes:
    print(f"Node {node}: {score:.6f}")

# Step 6: Visualize the graph
pos = nx.spring_layout(G, seed=42)

node_sizes = [1000 + 8000 * pr[node] for node in G.nodes()]

nx.draw_networkx(
    G,
    pos,
    with_labels=True,
    node_size=node_sizes,
    node_color="skyblue",
    edge_color="gray",
    arrows=True,
    font_size=10
)

plt.title("PageRank Algorithm")
plt.axis("off")
plt.show()

# Step 7: Verify PageRank scores
print("\nSum of PageRank scores:", round(sum(pr.values()), 6))

# Conclusion
print("\nConclusion: PageRank scores were calculated successfully.")
```
